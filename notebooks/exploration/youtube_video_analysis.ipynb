```python
import pandas as pd
import os
import config
from sqlalchemy import create_engine
import isodate
```

```python
# Create the database URL
DATABASE_URL = f'{config.DATABASE_TYPE}+{config.DBAPI}://{config.USER}:{config.PASSWORD}@{config.ENDPOINT}:{config.PORT}/{config.YOUTUBE_DATABASE}'

# Create the engine
engine = create_engine(DATABASE_URL)
```

```python
query = '''
SELECT * 
FROM "VIDEO_STATISTICS"

'''

# Read the table into a DataFrame
vaushYT_df = pd.read_sql(query, engine)
vaushYT_df.head()
```

```python
# Function to convert ISO 8601 duration to minutes
def convert_to_minutes(duration):
    parsed_duration = isodate.parse_duration(duration)
    total_minutes = parsed_duration.total_seconds() / 60
    return total_minutes

# Apply the function to the Duration column
vaushYT_df['DURATION_MINUTES'] = vaushYT_df['DURATION'].apply(convert_to_minutes)
```
